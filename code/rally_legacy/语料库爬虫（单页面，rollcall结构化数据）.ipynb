# 语料库爬虫

```
import os
import urllib.request
from bs4 import BeautifulSoup

url = input()
response = urllib.request.urlopen(url)
html_Text = response.read()
soup = BeautifulSoup(html_Text,'html.parser')
```

```
# 从 <title> 标签中提取地名和日期
title_element = soup.title
if title_element:
    title_text = title_element.get_text()
    # 假设标题格式是固定的，我们可以使用字符串分割来提取所需的部分
    title_parts = title_text.split(" - ")
    if len(title_parts) >= 3:
        location = title_parts[1].split(" in ")[1].split(" - ")[0]
        file_name = f"{location} - {title_parts[2]}.txt"
        file_path = os.path.join("C:\\Users\\numinose\\Desktop\\Trump rally\\test", file_name)

        # 确保目录存在
        os.makedirs(os.path.dirname(file_path), exist_ok=True)

        # 找到所有包含 "Donald Trump" 文本的 h2 元素
        h2_elements = soup.find_all("h2", text="Donald Trump")

        # 遍历所有 h2 元素，并提取紧随其后的 div 元素的文本
        content_to_write = ""
        for h2 in h2_elements:
            div_element = h2.find_next("div", class_="flex-auto text-md text-gray-600 leading-loose")
            if div_element:
                content_to_write += div_element.text + "\n"

        # 打开文件并写入内容
        with open(file_path, "w", encoding="utf-8") as file:
            file.write(content_to_write)

        print(f"文件已保存：{file_path}")

    elif len(title_parts) == 2:
        location = title_parts[0].split(" in ")[1].split(" - ")[0]
        file_name = f"{location} - {title_parts[1]}.txt"
        file_path = os.path.join("C:\\Users\\numinose\\Desktop\\Trump rally\\test", file_name)

        # 确保目录存在
        os.makedirs(os.path.dirname(file_path), exist_ok=True)

        # 找到所有包含 "Donald Trump" 文本的 h2 元素
        h2_elements = soup.find_all("h2", text="Donald Trump")

        # 遍历所有 h2 元素，并提取紧随其后的 div 元素的文本
        content_to_write = ""
        for h2 in h2_elements:
            div_element = h2.find_next("div", class_="flex-auto text-md text-gray-600 leading-loose")
            if div_element:
                content_to_write += div_element.text + "\n"

        # 打开文件并写入内容
        with open(file_path, "w", encoding="utf-8") as file:
            file.write(content_to_write)

        print(f"文件已保存：{file_path}")
    else:
        print("无法从标题中提取地名和日期")
else:
    print("没有找到 <title> 标签")
```
